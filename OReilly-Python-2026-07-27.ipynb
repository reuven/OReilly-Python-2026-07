# What are functions?

We've been using functions from the first day of this course. Functions, of course, are the *verbs* in Python. They do things:

- `print` (displays on the screen)
- `input` (waits to get input from the user, and turns that into a string)
- `len` (returns the number of elements in a string, list, tuples, or dict
- `sum` (which returns the total of a list of integers)
- `type` (returns the type of value that it was given)

We don't need these functions! We don't really need *any* functions, or almost any functions. However, it's really a pain to program without them. That's because they provide us with *abstraction*. 

The idea of abstraction is that we can take many small details and put them under a single name, and then forget about the details. We can use that name to think, to communicate, and even to build larger systems.

A function takes many different things that you want to do, and wraps them up under one name. That allows us you use that name, and that functionality, in other contexts, and to communicate with others about what you're doing.

Another reason to use functions, besides abstraction, is the DRY rule that we've discussed a bit -- don't repeat yourself! If you have the same code in several places in your program, then you almost certainly would be better off defining a function, and then invoking the function multiple times. That makes it easier for you to write your program, but also to read (understand) and also to debug and maintain it down the road.

# Let's define a function!

To define a function, we need to do a few things:

1. We use the keyword `def` , which means, "I want to define a new function."
2. Immediately after the word `def`, we give the function a name. This is actually a variable assignment -- the variable will be assigned a function value!
3. We then put `()` after the function name. Right now, these will be empty, but we will soon put *parameter* names inside, so that our function can take arguments.
4. At the end of the line, we put a `:`
5. Then we have an indented *function body*. In a function body, you can have any code you want -- `print`, `input`, `for`, `if`

The function body can be as long or short as you want. However, it's usually a good idea to keep the function body as short as possible, for easier understanding and maintenance.

The function doesn't execute when you define the function. Rather, it only executes when you put `()` after its name, once it has been defined.

```python
def hello():            # function definition line, with def, the function name, (), and :
    print('Hello!')     # a one-line function body, which just prints
```

```python
# how can I execute this new verb (function) that I taught Python? With ()

hello()  
```

```python
# what happens if I don't use ()?

hello    # I get back the function, without executing it -- I have instructions, but I'm not following them
```

# Exercise: Calculator

1. Define a function, `calc`, that when invoked asks three questions:
    - What is the first number?
    - What is the operator?
    - What is the second number?
2. Assign the response to each question to a variable (`first`, `op`, `second`)
3. If the numbers can be turned into integers, and if the operator is either `+` or `-`, then get the result of the calculation and print it, along with the inputs.
4. If the numbers cannot be turned into integers, scold the user.
5. If the operator is neither `+` nor `-`, then still print something, but say `no result` or the like.

Examples:

    calc()
    First: 10
    Operator: +
    Second: 30
    10 + 30 = 40

    calc()
    First: 10
    Operator: /
    Second: 30
    10 / 30 = (no result)

    calc()
    First: 10
    Operator: +
    Second: hello
    You must enter integers!

```python
def calc():
    first = input('First: ').strip()
    op = input('Operator: ').strip()
    second = input('Second: ').strip()

    # if we can turn first + second into integers,
    # then do so and print appropriate output
    if first.isdecimal() and second.isdecimal():
    
        first = int(first)
        second = int(second)
    
        if op == '+':
            result = first + second
        elif op == '-':
            result = first - second
        else:
            result = '(No result)'
    
        print(f'{first} {op} {second} = {result}')

    # if we cannot turn first and second into integers,
    # scold the user
    else:
        print(f'You did not provide numbers for {first} and/or {second}.')

calc()
```
