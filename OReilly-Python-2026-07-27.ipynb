# What are functions?

We've been using functions from the first day of this course. Functions, of course, are the *verbs* in Python. They do things:

- `print` (displays on the screen)
- `input` (waits to get input from the user, and turns that into a string)
- `len` (returns the number of elements in a string, list, tuples, or dict
- `sum` (which returns the total of a list of integers)
- `type` (returns the type of value that it was given)

We don't need these functions! We don't really need *any* functions, or almost any functions. However, it's really a pain to program without them. That's because they provide us with *abstraction*. 

The idea of abstraction is that we can take many small details and put them under a single name, and then forget about the details. We can use that name to think, to communicate, and even to build larger systems.

A function takes many different things that you want to do, and wraps them up under one name. That allows us you use that name, and that functionality, in other contexts, and to communicate with others about what you're doing.

Another reason to use functions, besides abstraction, is the DRY rule that we've discussed a bit -- don't repeat yourself! If you have the same code in several places in your program, then you almost certainly would be better off defining a function, and then invoking the function multiple times. That makes it easier for you to write your program, but also to read (understand) and also to debug and maintain it down the road.

# Let's define a function!

To define a function, we need to do a few things:

1. We use the keyword `def` , which means, "I want to define a new function."
2. Immediately after the word `def`, we give the function a name. This is actually a variable assignment -- the variable will be assigned a function value!
3. We then put `()` after the function name. Right now, these will be empty, but we will soon put *parameter* names inside, so that our function can take arguments.
4. At the end of the line, we put a `:`
5. Then we have an indented *function body*. In a function body, you can have any code you want -- `print`, `input`, `for`, `if`

The function body can be as long or short as you want. However, it's usually a good idea to keep the function body as short as possible, for easier understanding and maintenance.

The function doesn't execute when you define the function. Rather, it only executes when you put `()` after its name, once it has been defined.

```python
def hello():            # function definition line, with def, the function name, (), and :
    print('Hello!')     # a one-line function body, which just prints
```

```python
# how can I execute this new verb (function) that I taught Python? With ()

hello()  
```

```python
# what happens if I don't use ()?

hello    # I get back the function, without executing it -- I have instructions, but I'm not following them
```

# Return values

When we invoke `input`, we can put it on the right side of `=` (assignment), because it *returns* a value. When you invoke `input`, you get a string back on which you can invoke `str.strip` or which can be assigned. 

`len` also returns a value, an integer.

What about our functions? What do they return?

So far, nothing. More specifically, they all return the special value `None`. If a function in Python does not explicitly say what it returns with the special `return` keyword, then the function returns `None`.

What can you return from a Python function? Any Python value whatsoever, without any restrictions.

When you invoke `return`, the function execution ends. So the returned value is returned to the caller, and if there was any code after the `return`, it is ignored. Usually, if you have more than one `return` in a function, that's because each is inside of an `if` block.

Generally speaking, you don't want a function to `print`. You do want the function to `return`. Why? Because a returned value can be printed. But a printed value cannot be assigned to a variable. By returning a value, rather than printing it, you give the caller the option of what to do.

```python
def hello(name):
    return f'Hello, {name}!'  # return isn't a function; you don't need to use (), and we normally don't.
```

```python
s = hello('Reuven')
```

```python
s
```

# How do they know?

When someone wants to invoke one of our function, how do they know -- how can they know? -- what types of arguments we expect them to pass? It's very nice that I can say, "Everyone should invoke `hello` with a string," but is that written somewhere? What if I'm working on a big project with many colleagues -- where can they look?

In Python, we standardize this with a "docstring," meaning a string at the start of the function, on the first line, that is not assigned to anyone. It contains the documentation for the function.

If you're thinking that that this is like a comment, that's partly true:

- Comments are for the coders who will maintain a function (your garage, for your car)
- Docstrings are for people who will invoke a function (the owner/driver for the car)

Docstrings are traditionally written as triple-quoted strings -- meaning we use `"""` and `"""`, allowing us to have multiple lines of text inside of the docstring.

The content of the docstring can be whatever you want, but I suggest always saying:

- Expects -- what arguments should we expect?
- Modifies -- what does this function change, if anything? (files, databases, network, mutable data)
- Returns -- what does this function return to the caller?

If you have all three of these in your docstring, you're doing great.

The first line of the docstring is often displayed as a headline or summary, so keep that in mind.

```python
def add(first, second):
    """a friendly function that adds two values, usually numbers
    
    - Expects: Two integers, although floats will also work. Passing two strings or other values that know how to use `+` is also OK, if frowned upon
    - Modifies: Nothing
    - Returns: The result of applying + to these two values
    """
    return first + second

add(12, 14)
```

# Keyword arguments

So far, when we have called a function, we have passed the arguments as *positional arguments*. This means that the arguments are assigned to parameters based on their positions -- the first argument to the first parameter, the second argument to the second parameter, etc.

But sometimes, we want to be a bit more explicit about the connection between arguments and parameters. To do this, we can invoke a function passing arguments in a different way, known as "keyword arguments."



```python
def add(x, y):
    return x + y

# positional arguments looks like this:
add(10, 5)  
```

```python
# keyword arguments look  like this:
add(x=10, y=5)
```

The difference with keyword arguments is that we (the caller of the function) dictate which argument will be assigned to which parameter. It's not based on order. It's based on what we say.

This can be useful for more explicit, easier-to-understand function calls. It's also good when (as we'll see in a bit) we have optional parameters. 

```python
# parameters:  x    y
# arguments:   10  5

add(10, 5)  
```

```python
# parameters:  x    y
# arguments:   10    5 

add(x=10, y=5)  
```

```python
# with keyword arguments, the order no longer matters
# parameters:  x    y
# arguments:   5    10 

add(y=10, x=5)  
```

Can you mix positional and keyword arguments in the same function call?

Yes, but the rule is: All positional arguments must come before all keyword arguments.

```python
add(5, y=10)   # this is fine -- positional before keyword
```

```python
add(5, 10)  # only positional
```

```python
add(x=5, y=10)  # only keyword
```

```python
# this won't work
add(x=5, 10)    # you cannot have keyword before positional arguments
```

# Default argument values

It can be annoying to invoke a function that takes 3 arguments, when 2 of them barely change. Why should you always have to pass all of them? Why not have two of them keep defaults, and only pass arguments to those parameters when needed?

This is pretty common, and Python has a solution for it: Default argument values.

The idea is pretty simple: When you define a function, you can specify that certain parameters have default argument values. If you don't pass an argument to that parameter, the default value is used.

```python
def add(x, y):
    return x + y

# parameters: x  y
# arguments:  10  3

add(10, 3)     
```

```python
# what happens if I invoke it with only 1 argument?

# parameters: x   y
# arguments:  10 

add(10)     
```

```python
# let's redefine add, such that y has a default argument value -- that if we don't get an argument for y, the default is used

def add(x, y=9):   # this isn't a keyword argument, because it's in our function definition, not in our function invocation
    return x + y

# parameters: x  y
# arguments:  10  3

add(10, 3)     
```

```python

# parameters:  x    y
# arguments:  10    9

add(10)   
```

Default argument values demonstrate that every parameter needs to be assigned a value from an argument. The difference is that here, there weren't enough arguments. Python looked at the function definition and retrieved the default for y that was set at definition time, and assigned it to `y`.

When you're in the function body, you can't really know which parmeters were assigned arguments from the caller and which were assigned from defaults.

# A few rules for default argument values

1. When a parameter has a default argument value, it's effectively optional. If the caller doesn't specify it in the invocation arguments, the default value will be used.
2. When you define the function, all mandatory parameters (i.e., those without defaults) come **BEFORE** all parameters with defaults.
3. You can have any value as as default argument, in theory. In practice, don't make mutable defaults -- that is, don't give lists or dicts as defaults.
4. If a function has a number of parameters with defaults, and you invoke the function with keyword arguments, you can "skip over" one or more parameters.

```python
def count_characters(text, to_count='aeiou', lowercase=False):
    total = 0

    if lowercase:                # do we need to force lowercase:
        text = text.lower()      # assign the result of str.lower() to text
    
    for one_character in text:          # go through text, one character at a time
        if one_character in to_count:   # is this a character we want to count?
            total += 1
    return total         

count_characters('this is a test sentence', 'rstuvwxyz')    
```

# Any number of arguments with `*args`

What if I want to write a function that can handle any number of arguments? We can't quite do that yet, but we do have a pretty good substitute -- just pass a list as an argument.

But sometimes, it feels clunky to do that. 

```python
def mysum(numbers):
    total = 0

    for one_number in numbers:
        total += one_number

    return total

mysum([10, 20, 30, 40, 50])    
```

Nothing is wrong with this! My function is fine, and it's totally OK to pass a list of numbers to a function.

But... maybe I want to invoke the function with those integers as arguments.

```python
mysum(10, 20, 30, 40, 50)   # look, no [] !
```

If we want, we can redefine our function to have `*args` ("splat" is computer speak because it looks like a bug after you squished it). 

Having a parameter `*args` means a few things:

- You can actually give it any name you want, but `args` is traditional
- The `*` before the name makes it special, and means: This parameter will contain a tuple. The tuple's values will be all of the positional arguments that no other parameter was assigned.
- You can only have one `*args` in your function.
- It must come at the end of the parameter list.

```python
def mysum(*numbers):   # now we accept any number of positional arguments
    total = 0

    for one_number in numbers:
        total += one_number

    return total

mysum(10, 20, 30, 40, 50)     # all of these are positional arguments, all are thus shoveled into *numbers.. then we can use them
```

# Exercise: `low_high`

1. Write a function, `low_high`, that takes any number of positional arguments, all integers.
2. We can assume we'll have at least one argument.
3. The function will return a two-element list with the lowest and then highest integers from the input arguments.
4. (It's possible for them to be the same.)

Example:

    low_high(10, 15, 5, 20, 30, 7)   # return [5, 30]


```python
def low_high(*args):
    # set them up with the first element
    lowest = args[0]
    highest = args[0]

    for one_number in args:
        if one_number < lowest:   # is one_number < the current lowest?
            lowest = one_number   # ... one_number should be the new lowest

        if one_number > highest:
            highest = one_number

    return [lowest, highest]
```

```python
low_high(10, 15, 5, 20, 30, 7)
```
