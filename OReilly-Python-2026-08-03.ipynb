# What are modules?

We've talked several times about the DRY rule ("don't repeat yourself!")

- If the same line is repeated in a program, several times in a row, you can use a loop to "DRY up" your code.
- If the same functionality is repeated in different parts of a program, you can use a function to "DRY up" your code.
- If the same functionality is repeated in *multiple programs*, then we use a *library*.

Every programming language supports libraries. The idea is that a library contains data and functions that we will want to use in more than one place. Then, when we want to use it, we can just "borrow" that data or function from the library in our program.

In Python, we call our libraries "modules" and "packages." 

- A module is a single file with data and functions that we can load
- A package is a folder/directory of modules

Modules in Python provide us with at least three types of functionality:

- They contain the data and functions that we might want to use in multiple programs.
- Modules provide us with *namespaces*, sort of like surnames for variables, to ensure that two programs that use the same variable don't collide.
- Packages allow us to distribute what we've done to other Python programmers, either in our own organization or elsewhere on the Internet.

# Let's use a module!

If I want to use a module, then I can use the `import` statement. This loads a module from disk, creates a new value in memory that represents the module, and creates a variable with the module's name. So if I say:

````python
import random
````

# A few things to notice

1. `import` is a statement. It is not a function, so you don't put `()` after its name. You cannot redefine it.
2. The name that we put after `import` is used in at least two ways: (1) to find the file that is then loaded, and (2) the name of the variable that we define.
3. Many programming languages have similar functionality to `import`, but in those languages, you indicate the filename that you want to load, often as a string. This is *not* the case in Python!

After this `import` statement has run, the `random` variable is defined, and it contains the module `random`, with all of its data and functionality.

````python
type(random)  # what kind of value does random refer to?
````

````python
# let's let the random module print itself out

random
````

Once a module is loaded, I can use any value or function that it defines. The syntax for doing this is 

    MODNAME.NAME

So if I want to use something in `random`, I'll say

    random.NAME

If it's a function, I'll want to use `()` after the name to run it. If it's data, then I'll just use it.

That raises the question -- how can I know what names are defined?

1. You can use `dir(random)` to get a list of strings back. These strings are the attributes (i.e., names after `.`) that you can use to retrieve things.
2. You can use `help(random)` to get its docstring -- just as we saw functions have docstrings, so do modules. They often include information about the names defined.
3. You can look at `docs.python.org`, and go to the `random` module's documentation page in the "standard library reference."

# Exercise: Guessing game

1. Generate a random `secret` number from the `random` module, between 0 and 100.
2. Ask the user, repeatedly, to guess the number.
    - If they guess right, say "You got it!" and exit from the program
    - If they guess too low, say, "Too low!"
    - If they guess too high, say "Too high!"
3. If they got it wrong, have them try again.
4. Print the number of guesses, if you can.

````python
import random

secret = random.randint(0, 100)  
counter = 0

while True:
    counter += 1
    text = input('Guess: ').strip()
    
    if not text.isdecimal():
        print(f'{text} is not a number; try again')
        continue

    guess = int(text)
    
    if guess == secret:
        print('You got it!')
        break
    elif guess < secret:
        print('Too low!')
    else:
        print('Too high!')

print(f'You got it in {counter} guesses.')           
````

# Other ways to use `import`

There are two other variations on `import` you should know about:

- `import MODNAME as ALIAS` -- for example, `import pandas as pd`. The module is still loaded, but the variable that is defined isn't the same as the module filename.
- `from MODNAME import NAME as ALIAS` -- this allows you to import a name from a module, but give that name an alias, rather than itself. This is typical if you're worried about namespace collisions.

````python
import random   # random is a variable!
from random import randint  # randint is a variable!
````

# Variable names and modules names and function names are all ... names

When we assign to a variable:

    x = 5

I'm saying that the name `x` refers to the integer value 5.

When we define a function:

    def hello():
        return 'Hello!'

I'm saying that the name `hello` refers to the function value that we defined here.

If I import a module:

    import random

the module name is referring to a module value.

In all three cases, we're assigning to an indentifier, to a name in the Python world. There is no difference between variable names, function names, and module names.

When we say

    from random import choice as ch

Here, we're telling Python to (a) load the random module, (b) we only want the choice name defined as a variable, but (c) actually, we want the variable name to be ch, referring to the function random.choice.

If you use `import` on a module more than once, all but the first time will be largely ignored. The variable will still be defined (or redefined), but the module will not be reloaded.

# What happens when you `import` a module?

It's very common to believe that importing the module means that Python gets itself ready, if and when we ever need to use names in that module. But actually, Python doesn't work that way. Ever.

# `__name__` and `'__main__'`

`__name__` is a special variable that is always defined in Python, no matter where you are. It reflects the current *namespace*, meaning the current context in which variables are defined. 

`__name__` can have one of two values:

- A string name, the name of the module that is currently running
- The string `'__main__'`, which means that we're in the first program that was run, not a module that it imported.

If we check `__name__` and find that it's equal to `'__main__'`, then we know that the current program executing is the primary program, not a module.

Why would I care? How is this useful?

If we can distinguish between when our module is being imported and when it's being run directly, we can have a module that's kind of two-faced -- it can be imported, but it also can run on its own.

````python
reload(mymod)
````

Many  many many modules are designed to be imported, and also run. When you run a module, its `__name__` == `'__main__'` kicks into action.

But you don't need a `main` function or a `main.py` file or anyof this. It's just a convenience. But a really big one, and a very popular one, too.

# PyPI

Once you find a project on PyPI that seems useful and interesting, how do you install it and use it in your program?

The traditional answer is `pip`, a program that runs *outside* of Python, downloads the package's file(s) from PyPI, and then makes them available in your program's `sys.path` directory, so that they can be imported.

The syntax is, on the command line, either

    pip install NAME

or 

    python -m pip install NAME

The longer, second one is guaranteed to work with fewer problems.    

The package is then installed into your Python's `site-packages` directory, which is in `sys.path`.
