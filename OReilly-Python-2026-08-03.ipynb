```python
# here, I think the issue is using str.split, which returns a list of strings

with open('textfile.txt') as f:          # open the file for reading
    for line in f:                       # go through each line in the file
        word = line.strip()              # string, without leading/trailing whitespace
        print(word)

        print('\tDid it match?')
    
        if word[-1] in '.!:;\'"':
            print(f'\tYes: {word}')
```

```python
# we can use str.strip in a new way
# by default, str.strip removes all whitespace from both sides
# but we can pass an argument, a string of characters we want to remove!


punctuation = '.!:;\'"'

with open('textfile.txt') as f:          # open the file for reading
    for line in f:                       # go through each line in the file
        word = line.strip()              # string, without leading/trailing whitespace
        if word[-1] in punctuation:
            print(f'\tYes: {word.strip(punctuation)}')  # this removes all punctuation from both sides
```

# What are modules?

We've talked several times about the DRY rule ("don't repeat yourself!")

- If the same line is repeated in a program, several times in a row, you can use a loop to "DRY up" your code.
- If the same functionality is repeated in different parts of a program, you can use a function to "DRY up" your code
